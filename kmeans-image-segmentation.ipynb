```python
import matplotlib as mpl 
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import cv2 as cv
```

```python
img=cv.imread("coins.png")
img=cv.cvtColor(img,cv.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis("off")
```

```python
image=mpl.image.imread("coins.png")
plt.imshow(image)
plt.axis("off")
# image=mpl.image.imread("Imgur.jpg")
```

```python
img.shape
```

```python
#flattenning data ,
X=img.reshape(-1,3)
X.shape
```

```python
kmeans=KMeans(n_clusters=2,n_init=10)
```

```python
kmeans.fit(X)
```

```python
segmented=kmeans.cluster_centers_[kmeans.labels_]
segmented=segmented.reshape(img.shape)
```

```python
plt.imshow(segmented/255,cmap="gray")
```

```python
kmeans=KMeans(n_clusters=10,n_init=10)
kmeans.fit(X)
```

```python
segmented=kmeans.cluster_centers_[kmeans.labels_]
segmented=segmented.reshape(img.shape)
plt.imshow(segmented/255)
```
